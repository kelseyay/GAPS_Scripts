```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python

directory = r"/home/kelsey/GAPS_TrackerEnergyDep/build/Muons_MCv300TL1/Proxy_Beta/"
file_path = os.path.join(directory)
mu_beta_gen = uproot.open(file_path+"HGenWeight.root")
mu_beta_proxy = uproot.open(file_path+"HProxyWeight.root")
mu_beta_rec = uproot.open(file_path+"HRecWeight.root")


def get_hist_data(hist):
    values = hist.values()
    edges = hist.axis().edges()
    return values, edges

mu_beta_gen_hist = mu_beta_gen['HBetaGen_Weight;1']
mu_beta_proxy_hist = mu_beta_proxy['HBetaProxy_Weight;1']
mu_beta_rec_hist = mu_beta_rec['HBetaRec_Weight;1']

mu_beta_gen_vals, mu_beta_gen_edges = get_hist_data(mu_beta_gen_hist)
mu_beta_proxy_vals, mu_beta_proxy_edges = get_hist_data(mu_beta_proxy_hist)
mu_beta_rec_vals, mu_beta_rec_edges = get_hist_data(mu_beta_rec_hist)


plt.figure(figsize=(9,6))
plt.yscale('log')  # Sets y-axis to log scale

plt.stairs(
    mu_beta_gen_vals,
    mu_beta_gen_edges,
    color="#2495E0",
    fill=False,
    alpha=1,
    linewidth=2,
    label="Mu v300 MC Gen Beta Weighted",
    zorder=3,
)

plt.stairs(
    mu_beta_rec_vals,
    mu_beta_rec_edges,
    color="#D90291",
    fill=False,
    alpha=1,
    linewidth=2,
    label="Mu v300 MC Rec Beta Weighted",
    zorder=3,
)

plt.stairs(
    mu_beta_proxy_vals,
    mu_beta_proxy_edges,
    color="#ECE224",
    fill=False,
    alpha=1,
    linewidth=2,
    label="Mu v300 MC Proxy Beta Weighted",
    zorder=3,
)
```
