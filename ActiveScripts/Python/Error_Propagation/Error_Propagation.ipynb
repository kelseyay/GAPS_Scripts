```python
import math
from sympy import symbols, Eq, solve, I
import numpy as np 
#Error propagation of beta in TOF of GAPS

dt = 400e-12
dx = 6e-2
dy = 6e-2/np.sqrt(12)
dz = 1e-2
c = 3.00e8

z = 0.93 #meters between UMB and CBE top
```

```python
#Submit values for beta, x, y, z distance traversed by particle
x = 100e-2
y = 100e-2
true_beta = 0.51

d = np.sqrt(x**2 + y**2 + z**2)
#print(d)
dd = (1/d)*np.sqrt((x*dx)**2 + (y*dy)**2 + (z*dz)**2)
#print(dd)
true_t = d/(true_beta*c)
#print(true_t)

print(f"Distance error contribution: {(1/c**2)*(dd/true_t)**2}")
print(f"Timing error contribution: {(1/c**2)*((-d*dt)/true_t**2)**2}")
print(f"Just timing error contribution: {(1/c)*((-d*dt)/true_t**2)}")

db = (1/c)*np.sqrt((dd/true_t)**2 + ((-d*dt)/true_t**2)**2)
print(db)
```
