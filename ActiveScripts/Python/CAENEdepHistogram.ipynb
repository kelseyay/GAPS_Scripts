```python
import matplotlib.pyplot as plt

name = "N6725_73 - CH 0 (1).txt"
nlines = len(open(name).readlines(  ))
Counts = [0]*nlines
Events = [0]*nlines
ylow = 0.1
yhigh = 1000
title = "Energy Deposition Spectrum"
```

```python
pltfile = open(name, "r")
muhits = 0
for i, x in enumerate(pltfile):
    tarray = x.strip().split(' ')
    Counts[i] = float(tarray[0])
    Events[i] = float(tarray[1].replace(',', ''))
    if Counts[i] > 300 and Counts[i] < 1200:
        muhits = muhits + Events[i]
        #print(muhits)

print(muhits)
```

```python
plt.figure(figsize=(6, 5),dpi=116)
plt.yscale('log')
plt.xlabel("Counts (ADC?)",fontsize=13)
plt.ylabel("Events",fontsize=13)
plt.ylim(ylow,yhigh)
plt.title(title,fontsize=14)

plt.plot(Counts, Events)    
#plt.plot(V, I, 'o', label = lets[j], color=cols[j]) #Can uncomment this if for whatever reason you'd like colors on your plot. 
pltfile.close()

#plt.legend(loc='upper left')
#plt.savefig(savename) #Save the file
plt.show() #Display the plot
```
