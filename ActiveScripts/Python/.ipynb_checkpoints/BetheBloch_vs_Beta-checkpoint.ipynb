```python
import math
import matplotlib.pyplot as plt
from sympy import symbols, Eq, solve, I
import numpy as np 
```

```python
#Seeking to understand the behavior of Bethe-Bloch at extremes. 

ylow = 1
yhigh = 30

#Let's do muons in silicon
z = 1 #protons or muons
Zeff = 14
Aeff = 28
rho = 2.33 #Density g/cm^2
L = 1.5 #Length of material region in cm
dx = 0.01
#dx = 0.0005 #When you're ready for a final calculation, dx = 0.0005 is adequately small #This is the step size in the array which gets you from 0 to the length of the thing you're in.
x = np.arange(0, L, dx)
b = symbols('b')
#m = 938.3 #proton mass #105.658 #Muon mass
m = 105.658
print("m is " + str(m))

beta = [0.05,0.1,0.2,0.4,0.6,0.8, 0.9,0.99,0.999,0.9999,0.99999]
#beta = [0.7,0.75,0.8,0.85,0.87,0.9,0.93,0.95,0.97,0.99]
#beta = [0.755,0.765,0.775,0.785,0.795,0.805,0.815,0.825,0.835,0.845,0.855,0.865,0.875,0.885,0.895,0.905,0.915,0.925,0.935,0.945,0.955,0.965,0.975,0.985,0.995]
#beta = [0.9, 0.95]
gamma = [1/math.sqrt(1 - x**2) for x in beta]
lterm = np.zeros(len(beta))
dE = np.zeros(len(beta))
dEbetasq = np.zeros(len(beta))
gbet = np.zeros(len(beta))
coeff = np.zeros(len(beta))

for i in range(len(beta)):
    delta = 2*math.log(gamma[i])
    dE[i] = 0.3071*z**2*(Zeff/Aeff)*(1/beta[i]**2)*(math.log(1.022 * gamma[i]**2 * beta[i]**2 / (16 * Zeff**0.9 * 10**-6) ) - beta[i]**2)
    dEbetasq[i] = 0.3071*z**2*(Zeff/Aeff)*(math.log(1.022 * gamma[i]**2 * beta[i]**2 / (16 * Zeff**0.9 * 10**-6) ) - beta[i]**2)
    coeff[i] = beta[i]/np.sqrt(math.log(1.022 * gamma[i]**2 * beta[i]**2 / (16 * Zeff**0.9 * 10**-6) ) - beta[i]**2)
    gbet[i] = gamma[i]*beta[i]
    lterm[i] = math.log(1.022 * gamma[i]**2 * beta[i]**2 / (16 * Zeff**0.9 * 10**-6)) - beta[i]**2
    print("Log term = " + str(lterm[i])) 
    print("(1/Beta^2) * Log term = " + str((1/beta[i]**2) * lterm[i])) 
    print("gbet = " + str(gbet[i]))
    print("delta = " + str(delta))
    print("dEbetasq = " + str(dEbetasq[i]))
    print("coeff = " + str(coeff[i]))
    print("dE no delta = " + str(dE[i]))
    print()

print(dEbetasq)
plt.plot(beta,dE)
```

```python
beta = np.arange(0.1, 1, 0.01)
#print(beta)
gamma = [1/math.sqrt(1 - x**2) for x in beta]
coeff_tof = np.zeros(len(beta))
coeff_tkr = np.zeros(len(beta))
ratio = np.zeros(len(beta))

Geff_tof = 1
Geff_tkr = 1

Ztof = 5.574;
Atof = 10.3;
rho_tof = 1.032 #Density g/cm^2
L_tof = 0.635 #Length of material region in cm

Ztkr = 14;
Atkr = 28.3;
rho_tkr = 2.33 #Density g/cm^2
L_tkr = 0.22 #Length of material region in cm


for i in range(len(beta)):
    coeff_tof[i] = Geff_tof * (Atof/Ztof) * (1/0.307) * (1/(rho_tof*L_tof)) / ( math.log(1.022 * gamma[i]**2 * beta[i]**2 / (16 * Ztof**0.9 * 10**-6)) - beta[i]**2 ) 
    coeff_tkr[i] = Geff_tkr * (Atkr/Ztkr) * (1/0.307) * (1/(rho_tkr*L_tkr)) / ( math.log(1.022 * gamma[i]**2 * beta[i]**2 / (16 * Ztkr**0.9 * 10**-6)) - beta[i]**2 )
    ratio[i] = coeff_tkr[i] / coeff_tof[i]
    #coeff[i] = 0.3071*(Zeff/Aeff)*(math.log(1.022 * gamma[i]**2 * beta[i]**2 / (16 * Zeff**0.9 * 10**-6) ) - beta[i]**2)
plt.plot(beta,coeff_tof,label="TOF Coefficient")
plt.plot(beta,coeff_tkr,label="TKR Coefficient")
plt.plot(beta,ratio,"-",label ="ratio")
plt.xlabel("Beta")
plt.ylabel("Multiplicative Coefficient")
plt.legend()
```
