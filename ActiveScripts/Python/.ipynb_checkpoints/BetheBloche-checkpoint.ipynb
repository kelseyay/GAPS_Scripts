```python
import math
from sympy import symbols, Eq, solve, I
#I'm going to try to use Python to do my Bethe-Bloche simple stopping calculation. 
#The purpose of this block of code is to see how Beta > 0.8 vs Beta > 0.9 vs B > 0.99 muons are stopped in our tracker.
#We're going to do a tiny-step by tiny-step dE/dx and calculate every interesting value (new Beta, new momentum, new velocity*lifetime)
#Units are MeV and we're using natural units

#Start by trying to replicate the Muons in Copper plot
Zeff = 29
Aeff = 63.5
rho = 8.96 #Density

T = 20 #Initial energy in MeV, might as well keep it to KE because you can't siphon off mass energy o-o;;
dx = 0.001 #This will be an array eventually
p = math.sqrt((T + 105.658)**2 - 105.658**2) #This implementation of momentum is correct

b = symbols('b')
equation = Eq(b**2/(1-b**2), p**2/105.658**2) 
solutions = solve(equation, b)
beta = solutions[1] #This implementation of beta is correct
gamma = 1/math.sqrt(1 - beta**2)
print("p is " + str(round(p,2)) + " MeV")
print("Beta is " + str(beta))

dE = 0.3071*(Zeff/Aeff)*(1/beta**2)*(math.log(1.022 * gamma**2 * beta**2 / (16 * Zeff**0.9 * 10**-6) ) - beta**2) #Units of MeV/(g/cm^2)?
print("dE/dx is " + str(dE))

#So far so good! We're getting a real Bethe-Bloche curve!
#Eventually this will be a for loop with baby delta(x) steps
#dE = 0.3071*(5/10.5)*(1/Beta^2)*(log(1.022*Gamma^2*Beta^2/I) - Beta^2) #Constants * z_muon^2 * Z_plastic or Si / A * (1/Beta^2 ) * (ln(2m_e c^2 gamma^2 * Beta^2 / I - Beta^2))
#Using an effective A and Z for plastic
#Can also use polystyrene for foam and air for between the TOF panels. Maybe later, start with the ez stuff. 
```

```python
import math
from sympy import symbols, Eq, solve, I
import numpy as np 
#I'm going to try to use Python to do my Bethe-Bloche simple stopping calculation. 
#The purpose of this block of code is to see how Beta > 0.8 vs Beta > 0.9 vs B > 0.99 muons are stopped in our tracker.
#We're going to do a tiny-step by tiny-step dE/dx and calculate every interesting value (new Beta, new momentum, new velocity*lifetime)
#Units are MeV and we're using natural units

#Start by trying to replicate the Muons in Copper plot
Zeff = 29
Aeff = 63.5
rho = 8.96 #Density g/cm^2
L = 1 #Length of material region
dx = 0.01
#dx = 0.0005 #When you're ready for a finaly calculation, dx = 0.0005 is adequately small #This is the step size in the array which gets you from 0 to the length of the thing you're in.
x = np.arange(0, L, dx)
b = symbols('b')

T = 40 #Initial energy in MeV, might as well keep it to KE because you can't siphon off mass energy o-o;;

for i in range(len(x)):
    if(T < 0):
        print("Particle has stopped at length " + str(dx * i))
        break
    p = math.sqrt((T + 105.658)**2 - 105.658**2) #This implementation of momentum is correct
    equation = Eq(b**2/(1-b**2), p**2/105.658**2) 
    solutions = solve(equation, b)
    beta = solutions[1] #This implementation of beta is correct
    gamma = 1/math.sqrt(1 - beta**2)
    if i % (L/(10*dx))  == 0:
        print("L is " + str(dx * i))
        print("T is " + str(round(T,2)))
        print("p is " + str(round(p,2)) + " MeV")
        print("Beta is " + str(round(beta,2)))
    dE = 0.3071*(Zeff/Aeff)*(1/beta**2)*(math.log(1.022 * gamma**2 * beta**2 / (16 * Zeff**0.9 * 10**-6) ) - beta**2) #Units of MeV/(g/cm^2)
    T = T - dE*rho*dx

if T > 0:
    print("Particle has exited material with:")
    print("T is " + str(round(T,2)))
    print("p is " + str(round(p,2)) + " MeV")
    print("Beta is " + str(round(beta,2)))
```

```python
#Silicon detector
import math
from sympy import symbols, Eq, solve, I
import numpy as np 
#I'm going to try to use Python to do my Bethe-Bloche simple stopping calculation. 
#The purpose of this block of code is to see how Beta > 0.8 vs Beta > 0.9 vs B > 0.99 muons are stopped in our tracker.
#We're going to do a tiny-step by tiny-step dE/dx and calculate every interesting value (new Beta, new momentum, new velocity*lifetime)
#Units are MeV and we're using natural units
#I should possibly implement air too. Treat foam like air. Also tie together the whole code and do different "levels" 
#Bethe-Bloche, at least in Helium gas is very, very significant, so let's go ahead and implement the air as nitrogen gas or something
#See Grupen chapter 8

#Let's do muons in silicon
Zeff = 14
Aeff = 28
rho = 2.33 #Density g/cm^2
L = 1.5 #Length of material region
dx = 0.01
#dx = 0.0005 #When you're ready for a final calculation, dx = 0.0005 is adequately small #This is the step size in the array which gets you from 0 to the length of the thing you're in.
x = np.arange(0, L, dx)
b = symbols('b')
m = 938.3 #proton mass #105.658 #Muon mass

T = 50 #Initial energy in MeV, might as well keep it to KE because you can't siphon off mass energy o-o;;

for i in range(len(x)):
    if(T < 0):
        print("Particle has stopped at length " + str(dx * i))
        break
    p = math.sqrt((T + m)**2 - m**2) #This implementation of momentum is correct
    equation = Eq(b**2/(1-b**2), p**2/m**2) 
    solutions = solve(equation, b)
    beta = solutions[1] #This implementation of beta is correct
    gamma = 1/math.sqrt(1 - beta**2)
    if i % (L/(10*dx))  == 0:
        print("L is " + str(dx * i))
        print("T is " + str(round(T,2)))
        print("p is " + str(round(p,2)) + " MeV")
        print("Beta is " + str(round(beta,2)))
    dE = 0.3071*(Zeff/Aeff)*(1/beta**2)*(math.log(1.022 * gamma**2 * beta**2 / (16 * Zeff**0.9 * 10**-6) ) - beta**2) #Units of MeV/(g/cm^2)
    T = T - dE*rho*dx

if T > 0:
    print("Particle has exited material with:")
    print("T is " + str(round(T,2)))
    print("p is " + str(round(p,2)) + " MeV")
    print("Beta is " + str(round(beta,2)))
```
