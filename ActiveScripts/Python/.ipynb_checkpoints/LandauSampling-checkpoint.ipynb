```python
#Thanks Senger, you're such a real one: https://github.com/SengerM/landaupy
import numpy as np
from landaupy import landau
import matplotlib.pyplot as plt

#MPV Attempt
import math
import matplotlib.pyplot as plt
from sympy import symbols, Eq, solve, I
import numpy as np 

#z = 1 #protons or muons
#Zeff = 5.57 #14 #Si 14 
#Aeff = 10.3 #28.1 #Si 28
#rho = 1.032 #2.33 #Silicon Density 2.33 g/cm^2
#L = 0.635 #0.22 #Si(Li) 0.22 #Active length #Length of material region in cm

z = 1 #protons or muons
Zeff = 5.57 #14
Aeff = 10.3 #28.1
rho = 1.032 #2.33 #Silicon #Density g/cm^2
L = 0.635 #0.22 #Active length #Length of material region in cm
ion = 16 * Zeff**0.9 * 10**-6

beta = 0.99
gamma = 1/math.sqrt(1 - beta**2)
eps = 0.15355*z**2*(Zeff/Aeff)*(rho*L)*(1/beta**2) #Okay so this is xi, not epsilon lmao

MPV = eps*(math.log( (1.022 * gamma**2 * beta**2)/ion ) + math.log(eps/ion) + 0.2 - beta**2)

print("eps = " + str(eps))
print(MPV)
```

```python
# 1. Define parameters (MPV and xi are physical parameters)
#x_mpv = MPV
#xi = 0.055

x_mpv = 2
xi = 1 #what the heck is xi? #Oh okay it is in fact the epsilong thing (see slides) and it determines the width of the distribution
#And how far the tail extends out. 
xi2 = 2

#Go ahead and prove this out by Landau-ing the silicon Landau distribution, use beta = 0.99 or something lol

# 2. Define the range of x values
# landaupy.sample can be used to get an idea of the distribution range, or simply define a range manually
x = np.linspace(x_mpv - 10*xi, x_mpv + 20*xi, 50)
x2 = np.linspace(x_mpv - 10*xi, x_mpv + 20*xi2, 50)
print(x)

# 3. Calculate the PDF
#What the heck is xi?
pdf_values = landau.pdf(x, x_mpv=x_mpv, xi=xi) #Check the documentation I guess haha
pdf_values2 = landau.pdf(x2, x_mpv=x_mpv, xi=xi2)

# 4. Plot the distribution
plt.figure(figsize=(8, 5))
plt.plot(x, pdf_values, 'b-', lw=2, label=f'Landau PDF (MPV={x_mpv}, xi={xi})')
plt.plot(x2, pdf_values2, 'b-', lw=2, label=f'Landau PDF (MPV={x_mpv}, xi={xi2})')


# Add labels, title, and grid

#plt.ylim(0.1, 10)
#plt.yscale('log')
plt.xlabel('Energy Loss (arbitrary units)')
plt.ylabel('Probability Density')
plt.title('Landau Distribution Plot using landaupy')
plt.grid(True)
plt.legend()

# Display the plot
plt.show()
```

```python
from landaupy import landau

z = 1 #protons or muons
Zeff = 14
Aeff = 28.1
rho = 2.33 #Silicon #Density g/cm^2
L = 0.22 #Active length #Length of material region in cm
ion = 16 * Zeff**0.9 * 10**-6

Nsamps = 1000
Nhits = 5
xlow = 0.3
xhigh = 2

beta = 0.9
gamma = 1/math.sqrt(1 - beta**2)
eps = 0.15355*z**2*(Zeff/Aeff)*(rho*L)*(1/beta**2) #Okay so this is xi, not epsilon lmao
MPV = eps*(math.log( (1.022 * gamma**2 * beta**2)/ion ) + math.log(eps/ion) + 0.2 - beta**2)

TrM = np.zeros(Nsamps)

for x in range(Nsamps): #If want to push it all out at once need another for loop for 1 through 10 hits required
    samples = landau.sample(x_mpv=MPV, xi=eps, n_samples=Nhits)
    samples.sort()
    #print(samples[0:math.floor(Nhits/2)])
    TrM[x] = np.mean(samples[0:math.floor(Nhits/2)])
    #print(x)

custom_bins = np.arange(xlow,xhigh,0.05)

plt.hist(TrM, bins = custom_bins)

# 3. Add labels and a title
plt.xlim(xlow,xhigh)
plt.ylim(1, Nsamps)
plt.yscale('log')
plt.xlabel('Energy Deposited [MeV]')
plt.ylabel('NEntries')
plt.title('Truncated Mean Dist. B = '+ f'{beta:.2f} ' + f'NHits = {Nhits} NEvents = {Nsamps}')

#plt.title('Truncated Mean Dist. B = '+ f'{beta:.2f}' + f' MPV = {MPV:.2f} xi = {eps:.2f} Nsamples = {Nsamps}')

print(beta)
print(eps)
print(MPV)
print(f'Truncated Mean Distribution Mean: {np.mean(TrM)}')

# 4. Display the plot
plt.show()
```

```python
from landaupy import landau

z = 1 #protons or muons
Zeff = 14
Aeff = 28.1
rho = 2.33 #Silicon #Density g/cm^2
L = 0.22 #Active length #Length of material region in cm
ion = 16 * Zeff**0.9 * 10**-6

Nsamps = 100
Nhits = 10
xlow = 0.3
xhigh = 2
custom_bins = np.arange(xlow,xhigh,0.02)
print(f'Nhits {Nhits}')
print(f'Nsamps {Nsamps}')

beta = [0.7,0.8, 0.9,0.95,0.99]
#MPV = eps*(math.log( (1.022 * gamma**2 * beta**2)/ion ) + math.log(eps/ion) + 0.2 - beta**2)

for b in beta:
    TrM = np.zeros(Nsamps)
    eps = 0.15355*z**2*(Zeff/Aeff)*(rho*L)*(1/b**2)
    gamma = 1/math.sqrt(1 - b**2)
    MPV = eps*(math.log( (1.022 * gamma**2 * b**2)/ion ) + math.log(eps/ion) + 0.2 - b**2)
    dE = (rho*L)*0.3071*(Zeff/Aeff)*(1/b**2)*(math.log(1.022 * gamma**2 * b**2 / (16 * Zeff**0.9 * 10**-6) ) - b**2) #Units of MeV/(g/cm^2)
    for x in range(Nsamps): #If want to push it all out at once need another for loop for 1 through 10 hits required
        samples = landau.sample(x_mpv=MPV, xi=eps, n_samples=Nhits)
        samples.sort()
        TrM[x] = np.mean(samples[0:math.floor(Nhits/2)])
    #plt.hist(TrM, bins = custom_bins)
    #plt.xlim(xlow,xhigh)
    #plt.ylim(1, Nsamps)
    #plt.yscale('log')
    #plt.xlabel('Energy Deposited [MeV]')
    #plt.ylabel('NEntries')
    #plt.title('Truncated Mean Dist. B = '+ f'{b:.2f} ' + f'NHits = {Nhits} NEvents = {Nsamps}')
    #plt.show()
    print(f'Beta: {b} MPV: {MPV:.2f} dE: {dE:.2f} Truncated Mean Distribution Mean: {np.mean(TrM):.2f}')

print('I am done')
#custom_bins = np.arange(xlow,xhigh,0.05)
#plt.hist(TrM, bins = custom_bins)
```
