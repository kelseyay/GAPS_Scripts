```python
#Solving a system of three equations at the end of this!
import math
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
import numpy as np 
```

```python
r = 29 #radius of the dicing saw
D = 0.45 #depth of the cut
X = np.sqrt(r**2 - (r - D)**2)
print(f"Overcut = {X:.3f}")

theta = 2*np.arcsin( np.sqrt(X**2 + D**2)/(2*r) )
print(f"theta = {theta:.3f}")

def system(vars):
    t1, t2, d = vars
    eq1 = t1 + t2 - theta
    eq2 = t1 - (2*np.arcsin( np.sqrt((X-1)**2 + (D-d)**2)/(2*r) ))
    eq3 = t2 - (2*np.arcsin( np.sqrt(1**2 + d**2)/(2*r) ))
    return [eq1, eq2, eq3]

initial_guess = [1.0,1.0,1.0]

solution = fsolve(system, initial_guess)
print(f"Solution: theta_1 = {solution[0]:.3f}, theta_2 = {solution[1]:.3f}, d = {solution[2]:.3f}")
```
